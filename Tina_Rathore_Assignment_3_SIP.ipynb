```python
import numpy as np
import matplotlib.pyplot as plt
from scipy  import stats
data = np.random.randn(1000)
mean=np.mean(data)
#median
median=np.median(data)
#mode
mode_result=stats.mode(data,keepdims=False)
mode=mode_result.mode
std_dev=np.std(data,ddof=1)
data_range=max(data)-min(data)
print("Mean:",mean)
print("Median:",median)
print("range",data_range)
print("Mode:",mode)
print("Standard Deviation:",std_dev)
# Plot the histogram
plt.hist(data, bins=101, color='skyblue', edgecolor='black')
plt.title("Histogram of Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

```

```python
import random
heads = 0
b_gt_4 = 0
for _ in range(10000):
    coin = random.choice(['H', 'T'])

    if coin == 'H':
        heads += 1
        die = random.randint(1, 6)
        if die > 4:
            b_gt_4 += 1
p1 = b_gt_4 / heads
p2 = 1
print("P(B > 4 | A = Heads):", round(p1, 4))
print("P(B ≤ 4 | A = Tails):", p2)
```

**Problem - 7: (Enter the problem of your choice)** \\
Write the code below.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambdas = [3, 5, 7]
x = np.arange(0, 20)

for lam in lambdas:
    samples = np.random.poisson(lam, 1000)
    plt.hist(samples, bins=np.arange(0, 20), density=True, alpha=0.6, color='orange', label='Simulated')
    pmf = poisson.pmf(x, lam)
    plt.plot(x, pmf, 'bo-', label='Theoretical PMF')
    plt.title(f'Poisson Distribution (λ = {lam})')
    plt.xlabel('k')
    plt.ylabel('Probability')
    plt.legend()
    plt.grid(True)
    plt.show()
```

**Problem - 8: (Enter the problem of your choice)** \\
Write the code below.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, skew
mu = 50       # Mean
sigma = 10    # Standard deviation
size = 10000  # Number of samples

data = np.random.normal(mu, sigma, size)
plt.hist(data, bins=50, density=True, alpha=0.6, color='skyblue', label='Histogram')
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
pdf = norm.pdf(x, mu, sigma)
plt.plot(x, pdf, 'r', label='PDF (normal curve)')
plt.title('Histogram and Normal PDF')
plt.xlabel('Value')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)
plt.show()
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
print("Skewness:", skew(data))
```
